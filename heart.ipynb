#LIST TO DATAFRAME

```python
#list with five records, each records has id,name,age
import pandas as pd
students_list=[
    [10,"ROSELINE",18],
    [20,"OBADIA",22],
    [30,"MATHVER",20],
    [40,"SHYRAH",35],
    [50,"MPENDA",23]
]

#change list to dataframe and create columns_names
df_list=pd.DataFrame(students_list,columns=["ID","Name","Age"])
df_list
```

                         A2-DICTIONARY TO DATAFRAME

```python
#Dictionary with 5 records,3 columns: ID,Name,Age
students_dict={
    "ID":[10,20,30,40,50],
    "Name":["ROSELINE","OBADIA","MATHVER","SHYRAH","MPENDA"],
    "Age":[18,22,20,35,23]
}

df_dict=pd.DataFrame(students_dict)
df_dict

#display columns_names and datatype
print(df_dict.columns)
print(df_dict.dtypes)
```

                      A3-SERIES TO DATAFRAME

```python
#create series
students_name=pd.Series(["FRANCE","PHILIPO","AMOS","MICHAEL","SAMWEL"])
age=pd.Series([20,18,16,14,12])
score=pd.Series([98,96,90,92,91])

#combine series to dataframe
df_Series=pd.DataFrame({
    "Student Name": students_name,
    "Age": age,
    "Score": score
})
df_Series
```
