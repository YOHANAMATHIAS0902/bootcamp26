7.DATA TRANSFORMATION

```python
df['Dropout'] = df['target'].apply(lambda x: 1 if x == 'Dropout' else 0)

df['Age_Group'] = pd.cut(df['Age at enrollment'], bins=[16,20,24,30,100],
                          labels=['17-20','21-24','25-30','31+'])

df['Approval_Rate_1st'] = (df['Curricular units 1st sem (approved)'] /
                            df['Curricular units 1st sem (enrolled)'].replace(0, np.nan)) * 100

df['Performance'] = pd.cut(df['Curricular units 1st sem (grade)'],
                            bins=[-0.1,10,12,14,20],
                            labels=['Poor','Average','Good','Excellent'])

df[['Dropout','Age_Group','Approval_Rate_1st','Performance']].head()
```

This creates four new analysis-ready variables: a binary Dropout flag (1/0) for easy averaging, an Age_Group category, an Approval_Rate_1st percentage showing how many enrolled units a student actually passed, and a Performance category based on grade bands. These derived variables are what the rest of the notebook analyzes.

10.BUSINESS KPI ANALYSIS

```python
total_students = df.shape[0]
total_dropouts = int(df['Dropout'].sum())
dropout_rate = df['Dropout'].mean() * 100
retention_rate = 100 - dropout_rate
avg_1st_sem_grade = df[grade_col].mean()
pct_scholarship = df['Scholarship holder'].mean() * 100
pct_tuition_ok = df['Tuition fees up to date'].mean() * 100

kpis = {
    'Total Students': total_students,
    'Total Dropouts': total_dropouts,
    'Dropout Rate (%)': round(dropout_rate, 2),
    'Retention Rate (%)': round(retention_rate, 2),
    'Avg 1st Sem Grade': round(avg_1st_sem_grade, 2),
    '% Scholarship Holders': round(pct_scholarship, 2),
    '% Tuition Up to Date': round(pct_tuition_ok, 2),
}
pd.DataFrame(kpis.items(), columns=['KPI', 'Value'])
```

4,424 total students, of whom 1,421 dropped out — a 32.12% dropout rate (and correspondingly a 67.88% retention rate).
Average 1st semester grade is only 10.64 (on what's presumably a 0–20 scale), which is fairly low overall — consistent with the earlier finding that a chunk of students (especially dropouts) scored near 0.
24.84% hold a scholarship.
88.07% are up to date on tuition, meaning about 12% are behind on payments — this could be worth checking as a possible dropout risk factor.

14.INTERPRETATION OF RESULTS

**Finding**: The overall dropout rate is 32.1%, meaning roughly 1 in every 3 students leaves before completing their studies. Dropout rises sharply with age — from about 20% for students aged 17-20 to over 55% for students aged 25-30, before slightly declining for the 31+ group. The clearest gap, however, is financial: students who are **not up to date on tuition** drop out at a rate close to 86%, compared to roughly 23% for students who are up to date. 


**Interpretation**: Financial hardship is the strongest factor associated with dropout in this dataset — far stronger than gender, which shows only a modest difference between Female and Male students. Age also matters, but likely as a proxy for other pressures (work, family responsibilities) that older, non-traditional students face while studying. 


**Business Implication**: The institution should treat unpaid tuition as an early-warning signal, not just an administrative issue. Students falling behind on payments are the group most likely to leave, and this is something the institution can act on proactively rather than after the fact. 

15.GENERATE BUSINESS INSIGHTS

**Observation:** Students not up to date on tuition fees drop out at ~86%, versus ~23% for those up to date a gap of over 60 percentage points.

  **Insight:** Financial difficulty is by far the single strongest predictor of dropout in this dataset.

  **Recommendation:** Introduce flexible payment plans, early financial counseling, or short-term hardship grants for students who fall behind on tuition, before they reach the point of leaving.



**Observation:** Dropout rises steadily with age, from ~20% (17-20) to over 55% (25-30), then eases slightly for 31+.

  **Insight:** Older students likely balance study with work or family obligations, making them more vulnerable to dropping out.

  **Recommendation:** Offer flexible scheduling (evening/part-time options) and targeted advising for non-traditional (25+) students.



**Observation:** The gender gap in dropout rate is comparatively small next to the tuition and age gaps.

  **Insight:** Gender is not a major driver of dropout on its own in this institution interventions 
  should not be gender-targeted, but should focus on financial and age-related risk instead.



- **Observation:** The average 1st-semester grade across all students is 10.6 (out of 20) modest, with dropout students scoring meaningfully lower (see Section 9).

  **Insight:** Academic performance in the very first semester is an early signal of dropout risk, alongside financial status.

  **Recommendation:** Combine tuition-status monitoring with early-semester grade checks to flag at risk students as early as possible.
